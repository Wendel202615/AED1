#### Aprendendo operações baiscas em python


```python
n1 = float(input('Digite o numero 1: '))
n2 = float(input('Digite o numero 2: '))

soma = n1 + n2
subtracao = n1 - n2
multiplicacao = n1 * n2
divisao = n1 / n2

print(f'soma: {soma}')
print(f'subtracao: {subtracao}')
print(f'multiplicacao: {multiplicacao}')
print(f'divisao: {divisao}')
```

```python
quantidade = int(input('Digite a quantidade de itens:  '))
por_caixa = int(input('Digite a quantidade por caixa: '))

caixa_completa = quantidade // por_caixa
produtos_sobrando = quantidade % por_caixa

print(f'Quantidade de caixas completas: {caixa_completa}')
print(f'Quantidade de produtos sobrando: {produtos_sobrando}')
```

```python
numero = int(input('Digite o numero: '))
if numero % 2 == 0:
  print(f'O numero {numero} é par')
else:
  print(f'O numero {numero} é impar')
```

```python
preco_original = float(input('Digite o preço original:  '))
desconto = float(input('Digite o a porcentagem de desconto (%): '))

if preco_original <= 0:
  print('Erro o preço tem que ser maior que 0')
elif desconto < 0 or desconto > 50:
  print('Preço invalido, digite um valor de 0 a 50 %')
else:
  valor_desconto = preco_original * desconto / 100
  preco_final = preco_original - valor_desconto
print(f'O valor do desconto é R$ {valor_desconto}')
print(f'O valor final é R$ {preco_final}')
```

```python
vendas = float(input('Digite o valor de vendas: '))

comissao = vendas * 0.04
print(f'A comissão é: {comissao}')

if vendas > 15000:
  print(f'Exelente trabalho, continue assim!!')

```

```python
vendas = float(input('Digite o valor de vendas: '))

comissao = vendas * 0.04
print(f'A comissão é: {comissao:.2f}')

if vendas > 15000:
    print('Exelente trabalho, continue assim!!')
```
