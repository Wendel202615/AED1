```python
for char in 'a,d,z,p,s'.replace(' ', '').split(','):
    print(f"ord('{char}'): {ord(char)}")

for num_str in '65,68,80,90,53'.split(','):
    num = int(num_str)
    print(f"chr({num}): {chr(num)}")
```
