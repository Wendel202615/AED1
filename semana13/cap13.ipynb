```python
palavra = input('Digite uma palavra:  ') .strip()
if len (palavra) == 0:
  print(f'Nenhuma palavra foi digitada')
else:
  print(f'primeiro caractere: {palavra[0]}')
  print(f'ultimo caractere: {palavra[-1]}')
  print(f'tamamho: {len(palavra)}')
```

```python
palavra = input('Digite uma palavra: ').strip()
invertida = palavra[::-1]
print(invertida)
```

```python
nome = input('Digite seu nome: ')
nome_padrao = nome.strip().title()
print(nome_padrao)
```

```python
cpf_entrada = input('Digite seu CPF:  ')
cpf = cpf_entrada.strip().replace('.','.').replace('-','-')
if cpf.isdigit() and len(cpf) == 11:
  print(f'CPF valido: {cpf}')
else:
  print(f'cpf invalido {cpf}')
```

```python
frase = input('Digite a frase: ')
palavras = frase.split()
print(f'Total de palavras: {(len(palavras))}')
```

```python
nome = input('Digite seu nome: ')
nome = nome.strip()
nome = nome.upper()
palavras = nome.split()
quantidade = len(palavras)
print(f'palavras: {palavras}')
print(f'a quantidade de palavras é {quantidade}')
```
